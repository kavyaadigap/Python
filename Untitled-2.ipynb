```python
import pandas as pd
from sklearn.datasets import fetch_california_housing

# Fetch the data
housing_data_raw = fetch_california_housing()
# Transform to a Pandas DataFrame
housing_df = pd.DataFrame(data=housing_data_raw.data, columns=housing_data_raw.feature_names)

# Simulate missing values for illustration purposes
import numpy as np
housing_df.loc[::100, 'MedInc'] = np.nan 

# Counting Missing Values
missing_values = housing_df.isnull().sum()
print("Missing values in MedInc before Deletion: ", missing_values['MedInc'])
missing_values.index
# TODO: Listwise deletion of any rows with missing values
columns_missing=list()
for col in housing_df.columns:
    if housing_df[col].isnull():
        columns_missing.append(col)
housing_df_complete = housing_df.dropna

# Counting Missing Values
missing_values = housing_df_complete.isnull().sum()
print("Missing values in MedInc after Deletion: ", missing_values['MedInc'])
```
